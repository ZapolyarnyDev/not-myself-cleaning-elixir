# Установка и развертывание

## Системные требования

### Обязательные компоненты

- **Elixir** 1.16 или выше
- **Erlang/OTP** 25 или выше
- **PostgreSQL** 13 или выше
- **Git** для клонирования репозитория

### Опциональные компоненты

- **Nix** для управления окружением разработки
- **Just** для упрощения команд

## Проверка установленных компонентов

```bash
# Проверка Elixir
elixir --version
# Ожидаемый вывод: Elixir 1.16.x (compiled with Erlang/OTP 25)

# Проверка PostgreSQL
psql --version
# Ожидаемый вывод: psql (PostgreSQL) 13.x

# Проверка Git
git --version
# Ожидаемый вывод: git version 2.x.x
```

## Установка проекта

### Шаг 1: Клонирование репозитория

```bash
git clone <repository-url>
cd not-myself-cleaning-elixir
```

### Шаг 2: Настройка переменных окружения

```bash
cp .env.example .env
```

Отредактируйте файл `.env`:

```bash
DATABASE_URL=postgres://username:password@localhost/not_myself_cleaning_dev
PORT=4000
```

Замените `username` и `password` на учетные данные PostgreSQL.

### Шаг 3: Установка зависимостей

```bash
mix deps.get
```

Эта команда загрузит все необходимые библиотеки из hex.pm.

### Шаг 4: Настройка базы данных

```bash
# Создание базы данных
mix ecto.create

# Применение миграций
mix ecto.migrate
```

## Запуск приложения

### Режим разработки

```bash
mix phx.server
```

Или с использованием just:

```bash
just server
```

Приложение будет доступно по адресу: **http://localhost:4000**

### Интерактивная консоль

```bash
iex -S mix phx.server
```

Запускает сервер с интерактивной консолью Elixir.

## Использование Nix (опционально)

Если в системе установлен Nix:

```bash
# Активация окружения через direnv
direnv allow

# Или вручную
nix develop
```

Nix автоматически установит все необходимые зависимости.

## Возможные проблемы и решения

### Ошибка: "could not connect to server"

**Причина:** PostgreSQL не запущен или неверные параметры подключения.

**Решение:**

```bash
# Проверка статуса PostgreSQL (Linux/macOS)
sudo systemctl status postgresql

# Запуск PostgreSQL
sudo systemctl start postgresql

# Проверка параметров в .env
cat .env
```

### Ошибка: "database does not exist"

**Причина:** База данных не создана.

**Решение:**

```bash
mix ecto.create
```

### Ошибка: "port 4000 already in use"

**Причина:** Порт 4000 занят другим процессом.

**Решение:**

```bash
# Изменить порт в .env
PORT=4001

# Или найти и остановить процесс
lsof -i :4000
kill -9 <PID>
```

### Ошибка: "mix: command not found"

**Причина:** Elixir не установлен или не добавлен в PATH.

**Решение:**

Установите Elixir согласно инструкциям на https://elixir-lang.org/install.html

## Полезные команды

### Just команды

Проект использует Just для упрощения команд. Список всех команд:

```bash
# Показать все доступные команды
just

# Полная инициализация проекта
just init

# Запуск сервера
just server

# Запуск с консолью
just console

# Форматирование кода
just format

# Проверка и тесты
just check

# Показать маршруты
just routes
```

### Управление базой данных

```bash
# Just команды
just db-create          # Создать БД
just db-migrate         # Применить миграции
just db-rollback        # Откатить последнюю миграцию
just db-reset           # Пересоздать БД
just db-status          # Статус миграций
just db-gen-migration name  # Создать новую миграцию

# Mix команды
mix ecto.create         # Создать базу данных
mix ecto.drop           # Удалить базу данных
mix ecto.migrate        # Применить миграции
mix ecto.reset          # Пересоздать базу данных
mix ecto.rollback       # Откатить последнюю миграцию
mix ecto.rollback --all # Откатить все миграции
mix ecto.migrations     # Показать статус миграций
```

### Разработка

```bash
# Just команды
just compile            # Компиляция
just clean              # Очистка
just test               # Тесты
just test-coverage      # Тесты с покрытием

# Mix команды
mix format              # Форматирование кода
mix format --check-formatted  # Проверка форматирования
mix test                # Запуск тестов
mix test --cover        # Тесты с покрытием
mix compile             # Компиляция
mix phx.routes          # Показать маршруты
```

### Управление зависимостями

```bash
# Just команды
just setup              # Установить зависимости
just deps-update        # Обновить все зависимости
just deps-outdated      # Показать устаревшие

# Mix команды
mix deps.get            # Установить зависимости
mix deps.update --all   # Обновить все
mix hex.outdated        # Устаревшие пакеты
```

### Очистка

```bash
# Очистить скомпилированные файлы
just clean
# или
mix clean

# Удалить зависимости
rm -rf deps _build
```

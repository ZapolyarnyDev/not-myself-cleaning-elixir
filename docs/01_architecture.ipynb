# Архитектура проекта Not Myself Cleaning

## Общая структура

Проект построен на основе Phoenix Framework и имеет несколько слоёв приложения

## Слои приложения

### 1. Бизнес-логика

Расположение: `lib/not_myself_cleaning/`

Контексты и схемы данных:

- **Accounts** - управление пользователями и сессиями
- **Requests** - управление заявками на уборку

### 2. Веб-слой (Web Layer)

Расположение: `lib/not_myself_cleaning_web/`

Контроллеры, представления и маршруты:

- **Controllers** - обработка HTTP-запросов
- **HTML Modules** - рендеринг шаблонов
- **Router** - определение маршрутов
- **Endpoint** - точка входа HTTP-запросов

## Структура директорий

```
lib/
├── not_myself_cleaning/
│   ├── accounts/
│   │   ├── user.ex              # Схема пользователя
│   │   └── session.ex           # Схема сессии
│   ├── requests/
│   │   └── cleaning_request.ex  # Схема заявки
│   ├── accounts.ex              # Контекст Accounts
│   ├── requests.ex              # Контекст Requests
│   ├── application.ex           # OTP Application
│   └── repo.ex                  # Ecto Repository
│
└── not_myself_cleaning_web/
    ├── controllers/
    │   ├── auth_controller.ex
    │   ├── request_controller.ex
    │   ├── admin_controller.ex
    │   ├── auth_html.ex
    │   ├── request_html.ex
    │   ├── admin_html.ex
    │   └── layouts.ex
    ├── router.ex
    └── endpoint.ex
```

## Контексты

### Accounts Context

Функции:
- `register_user/1` - регистрация нового пользователя
- `authenticate_user/2` - аутентификация по логину и паролю
- `create_session/1` - создание сессии
- `get_session/1` - получение сессии по токену
- `delete_session/1` - удаление сессии

### Requests Context

Функции:
- `list_all_requests/0` - получение всех заявок
- `list_user_requests/1` - получение заявок пользователя
- `create_request/2` - создание новой заявки
- `update_request_status/3` - обновление статуса заявки

## Схемы данных

### User

Поля:
- `id` (UUID) - первичный ключ
- `login` (string) - уникальный логин
- `password` (string) - хешированный пароль
- `full_name` (string) - полное имя
- `phone` (string) - телефон
- `email` (string) - электронная почта

### CleaningRequest

Поля:
- `id` (UUID) - первичный ключ
- `user_id` (UUID) - внешний ключ на users
- `full_name` (string) - имя клиента
- `phone` (string) - телефон
- `email` (string) - email
- `address` (string) - адрес
- `contact` (string) - контактные данные
- `service` (string) - тип услуги
- `date_time` (string) - дата и время
- `payment` (string) - способ оплаты
- `status` (string) - статус заявки
- `status_comment` (string) - комментарий к статусу
- `created_at` (datetime) - дата создания

### Session

Поля:
- `token` (string) - первичный ключ
- `user_id` (string) - ID пользователя
- `login` (string) - логин
- `full_name` (string) - полное имя
- `phone` (string) - телефон
- `email` (string) - email
- `is_admin` (boolean) - флаг администратора
- `created_at` (datetime) - дата создания

## Контроллеры

### AuthController

Маршруты:
- `GET /` - редирект на /login или /requests
- `GET /register` - страница регистрации
- `POST /register` - обработка регистрации
- `GET /login` - страница входа
- `POST /login` - обработка входа
- `POST /logout` - выход из системы

### RequestController

Маршруты:
- `GET /requests` - список заявок пользователя
- `GET /requests/new` - форма создания заявки
- `POST /requests` - создание заявки

### AdminController

Маршруты:
- `GET /admin` - список всех заявок
- `POST /admin/status` - обновление статуса заявки
